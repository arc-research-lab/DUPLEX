```python
import sys
sys.path.insert(0, "..")          # repo root, so `dse` and `workloads` import

import dse.duplex_dse as dd
from workloads import load_suite

TAG = "spmv"
workloads = load_suite(TAG)
DEPTH_CONFIGS = [(4, 4)]
USE_CACHE = True

for name, (x, y, V, w) in workloads.items():
    print(f"{name:20s} reads={len(x):>7d}  V={V}")
print(f"L={dd.L}  BUDGET={dd.BUDGET}  N_TOTALS_A={dd.N_TOTALS_A}  N_TOTALS_B={dd.N_TOTALS_B}")
```

```python
depth_df = dd.run_depth_sweep(workloads, DEPTH_CONFIGS, TAG)
depth_df.pivot_table(index="D", columns=["workload", "m_w", "n_w"], values="eff")
```

```python
sims = dd.make_sims(workloads, TAG, use_cache=USE_CACHE)
target_best = dd.run_dse_target(workloads, sims, TAG)
dd.save_sims(sims, TAG)
target_best
```

```python
budget_best = dd.run_dse_budget(workloads, sims, TAG)
dd.save_sims(sims, TAG)
budget_best
```

```python
baseline_df = dd.run_baselines(workloads, sims, target_best, budget_best, TAG)
dd.save_sims(sims, TAG)
baseline_df
```
